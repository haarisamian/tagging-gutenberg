```python
from booknlp.booknlp import BookNLP
from pathlib import Path

params = {
    "pipeline": "entity,quote,supersense,event,coref",
    "model": "big",
    "device": "cuda"
}

bnlp = BookNLP("en", params)

input_dir = Path("/local/nlp/ham2176/gutenberg/data/chapters")
output_dir = Path("/local/nlp/ham2176/gutenberg/booknlp_out")
output_dir.mkdir(parents=True, exist_ok=True)

for txt_file in sorted(input_dir.glob("*.txt")):
    print(f"Processing {txt_file.name} ...")
    bnlp.process(str(txt_file), str(output_dir), txt_file.stem)
print("Finished processing all chapters!")
```
